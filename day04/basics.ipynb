Real programs deal with lots of data:
- A list of students
- A shopping cart
- A dictionary of word definitions
- A set of unique tags

Storing each in its own variable is impossible:

student1 = "Alice"
student2 = "Bob"
student3 = "Charlie"
what if there are 10,000 students?

Collections solve this. A collection is a single object that holds many values, and you can loop over it, search it, sort it,and grow it.

Python gives you FOUR main collection types: lists, tuples, sets, dictionary

Ordered  → items keep their position (you can index them) → lists, tuples, dictionary
Mutable  → you can change it after creating it → lists, sets, dictionary         
Duplicates → can the same value appear more than once? → lists, tuples, dictionary (key unique)


### LISTS
A list is an ORDERED, MUTABLE sequence of items, written in square brackets, separated by commas. Items can be ANY type, and can even be mixed (though mixing is usually a bad idea).
