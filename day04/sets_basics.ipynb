#### SETS — UNORDERED & UNIQUE
A set is an UNORDERED collection of UNIQUE items. No duplicates, no indexing (because there's no order). Written with curly braces.

```python
# --- Creating sets & removing duplicates ---
fruits = {"apple", "banana", "cherry", "banana"}
print(fruits)

fruits = {"apple", "banana", "apple", "cherry"}
print(fruits)
```

```python
# strip duplicates from a list.
names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
unique = set(names)
print(unique)
```

```python
# --- Adding and removing ---
s = {1, 2, 3}
s.add(4)                   # add one item
print(s)                   # {1, 2, 3, 4}

s.add(2)                   # adding a duplicate does nothing
print(s)                   # {1, 2, 3, 4}  (still no duplicate)

s.remove(3)                # remove (ERRORS if missing)
print(s)                   # {1, 2, 4}

s.discard(99)              # remove (NO error if missing)
print(s)                   # {1, 2, 4}
```

```python
# --- Set operations (math!) ---
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)                  # union:        {1,2,3,4,5,6}
print(a & b)                  # intersection: {3, 4}
print(a - b)                  # difference:   {1, 2}
print(b - a)                  # difference:   {5, 6}
print(a ^ b)                  # symmetric diff: {1,2,5,6}

# Sets are great for "which items are in BOTH lists?" or
# "which items are in A but not B?" — membership is very fast.
```
