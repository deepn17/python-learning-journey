```python
# Challenge 1 — Profit calculator
# Build a small business calculation that determines profit from revenue and cost, while keeping the result usable by downstream software and allowing a display currency to be specified.
# Business requirements
# •	Calculate profit as revenue minus cost.
# •	Accept revenue and cost as required inputs.
# •	Require currency as a keyword-only input so callers state it explicitly.
# •	Keep calculation output numeric; formatting is a presentation concern.

#technical requirements
#•	Implement `calculate_profit(revenue, cost, *, currency)`.
#•	Validate numeric inputs and reject bool values even though bool is a subclass of int.
#•	Return a numeric result rather than mixing calculation and display formatting.
#•	For financial-grade arithmetic, consider Decimal and a documented rounding policy.

# Notes for self
# Function Signature Design - (a, b, *, c) - The * means anything on the left of the star is positional or keyword arguments and anything right to the * is keyword only arguments.
# use isinstance for determining the type of the input

```

```python
# Solution
from decimal import Decimal
def calculate_profit(revenue, cost, *, currency):
    """
    Calculate profit from revenue and cost using Decimal arithmetic.

    Business Purpose:
        Calculate the net profit or loss for a financial transaction
        while maintaining decimal precision.

    Parameters:
        revenue (int, float, Decimal):
            Total revenue generated.

        cost (int, float, Decimal):
            Total cost incurred.

        currency (str):
            Currency code supplied as a keyword-only argument.
            Examples: INR, USD, EUR.

    Returns:
        Decimal:
            Calculated profit or loss.

    Raises:
        TypeError:
            If revenue or cost is not numeric, or if either
            input is a Boolean value.

    Design Decisions:
        1. Currency is keyword-only to make the function call explicit.
        2. Boolean values are rejected because bool is a subclass of int.
        3. Float values are converted through str() before Decimal
           conversion to avoid binary floating-point artefacts.
        4. All calculations are performed using Decimal.
        5. Only the calculated result is returned. Currency formatting
           is handled by the calling application.
    """
    # STEP 1: Validate revenue.
    # Accept int, float and Decimal.
    # Explicitly reject bool because bool inherits from int.

    if not isinstance(revenue, (int, float, Decimal)) or isinstance(revenue, (bool)):
        raise TypeError(f"Revenue must be numeric.")
    # STEP 2: Normalize revenue.
    # Convert float through str() to avoid binary floating-point
    # representation issues when creating a Decimal.
    elif isinstance(revenue, (float)):
        revenue = Decimal(str(revenue))
    
    # STEP 3: Validate cost using the same numeric contract.
    if not isinstance(cost, (int, float, Decimal)) or isinstance(cost, (bool)):
        raise TypeError(f"Cost must be numberic")
    
    # STEP 4: Normalize cost.
    elif isinstance(cost, (float)):
        cost = Decimal(str(cost))
    
    # STEP 5: Perform the business calculation.
    # Convert integer inputs to Decimal as well, ensuring that
    # subtraction is performed consistently using Decimal.
    result =  Decimal(revenue) - Decimal(cost)

    # STEP 6: Return the numeric result.
    # Do not format currency here; presentation is the caller's responsibility.
    return result

```

```python
# Test cases # 1
print(calculate_profit(1000, 700, currency="INR"))
```
