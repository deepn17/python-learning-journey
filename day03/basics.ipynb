### Day 3: Loops — for / while

Without loops, printing 1 to 5 looks like this:
    print(1)
    print(2)
    print(3)
    print(4)
    print(5)

with a loop, it's just:
    `for i in range(1, 6):
        print(i)`

Loops let you write the instruction ONCE and run it MANY times.

##### THE for LOOP — ITERATE OVER A SEQUENCE

`Syntax:
   for item in sequence:
       indented block — runs once per item`

The for loop visits each element of a "sequence" (string, list, range, etc.) one at a time, binding it to the loop variable.


```python
# Loop over a string
word = "Python"
for letter in word:
    print(letter)

# Each character of "Python" is printed on its own line.
# The loop variable (letter) takes a new value each iteration.
```

#### THE range() FUNCTION
range() generates a sequence of integers. Three forms:

*   range(stop)            → 0, 1, 2, ..., stop-1
*   range(start, stop)     → start, start+1, ..., stop-1
*   range(start, stop, step) → start, start+step, ..., before stop

Key Rule: stop is EXCLUSIVE — it is never included.

```python
# Example: 3 forms of range

print("range(5)   →", list(range(5)))
print("range(2, 7) →", list(range(2, 7)))
print("range(1, 10, 2) →", list(range(1, 10, 2)))
print("range(10, 0, -1) →", list(range(10, 0, -1)))

# list() converts a range into a visible list
```

```python
# Example: Countdown
for i in range(5, 0, -1):
    print(i)
print("Take Off")

# Negative step counts DOWN. Note: stop (0) is still exclusive.
```

#### while LOOP — REPEAT WHILE A CONDITION IS TRUE

`Syntax:
   while condition:
       indented block — runs as long as condition is True`

Use for when you know HOW MANY times to loop (a range).
Use while when you DON'T know — you loop until something changes.

```python
#Example: Count up with while
count = 1
while count <= 5:
    print(f"Count: {count}")
    count += 1                 # update the variable or infinite loop!

# If you forget `count += 1`, count stays 1 forever → INFINITE LOOP.
# Press Ctrl+C in the terminal to stop a runaway program.
```

```python
# Example: while for unknown iterations

# Keep asking until the user gives a valid answer.
answer = ""
while answer != "yes":
    answer = input("Type 'yes' to continue: ").lower().strip()
print("Thank you.")

# We don't know how many tries it'll take — perfect for while.
```

#### break AND continue — CONTROLLING THE LOOP
break    → exit the loop IMMEDIATELY
continue → skip the REST of this iteration, go to the next one

```python
#Example: break
for i in range(1, 11):
    if i == 6:
        print("Found 6 - stopping")
        break
    print(i)

# Prints 1..5, then stops at 6. 7..10 never run.
```

```python
# Example: continue
for i in range(1, 11):
    if i % 2 == 0:
        continue         # skip even numbers
    print(i)

# Prints only odd numbers: 1, 3, 5, 7, 9.
```

#### THE LOOP else CLAUSE
A loop can have an else block. It runs ONLY if the loop finished WITHOUT hitting a break.

```python
# Example: Prime number check
n = int(input("Enter a number: "))

for i in range(2, n):
    if n % i == 0:
        print(f"{n} is not prime (divisible by {i}).")
        break
else:
    print(f"{n} is prime!")

# If we break (found a divisor), else is skipped.
# If the loop completes fully, else runs → it's prime.
```

```python
# NESTED LOOPS — LOOPS INSIDE LOOPS
# Example: Multiplication table

for i  in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print("-" * 15)

# The inner loop runs COMPLETELY for each pass of the outer loop.
# Total iterations = outer × inner = 3 × 3 = 9.
```

```python
#Example: Pattern printing
for i in range(1, 6):
    print("*" * i)
```

```python
# ACCUMULATOR PATTERN — BUILDING A RESULT
# The most common loop pattern: start with an empty result, add to it each iteration.

#Example: Sum of numbers
total = 0            # accumulator starts at 0
for i in range(1, 6):
    total += i       # add each number
print(f"Sum of 1...5 = {total}")
```

```python
#Example : Building a string
result = ""                               # empty string accumulator
for letter in "hello":
    result += letter.upper()
print(result)
```

```python
# LOOPING WITH INDEX — enumerate()
# Sometimes you need BOTH the item AND its position.
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

# enumerate() gives (index, item) pairs — very common in real code.
```

```python
# Example: Input validation loop
while True:
    age = input("Enter your age: ").strip()
    if age.isdigit() and 0 < int(age) < 150:
        print(f"Valid age: {age}")
        break
    print("Invalid age. Try again")

# while True + break is the idiomatic "loop until valid" pattern.
```

```python
# Example : Menu loop
while True:
    print("\n1. Say hello")
    print("2. Say goodbye")
    print("3. Quit")
    choice = input("Choose: ").strip()

    if choice == "1":
        print("Hello")
    elif choice == "2":
        print("Goodbye")
    elif choice == "3":
        print("Existing")
        break
    else:
        print("Invalid choice")
```

```python
# Example: Flag-controlled loop
running = True
while running:
    cmd = input("> ").lower().strip()
    if cmd == "stop":
        running = False
    elif cmd == "":
        continue
    else:
        print(f"You said: {cmd}")
print("Loop ended.")
```
