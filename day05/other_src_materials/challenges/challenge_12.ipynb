```python
# Challenge 12 — Reusable operation registry and dynamic registration
#Build an extensible operation registry so operations can be registered and later selected by name without hardcoding every operation into the execution function.
#Business requirements
# •	Support registering named operations.
# •	Allow a caller to execute a registered operation dynamically.
# •	Reject unsupported operation names clearly.
# •	Keep registration reusable and separate from execution.
# Technical requirements
# •	Define business operation functions first.
# •	Initialize a registry dictionary.
# •	Implement reusable `register_operation` that validates name and callable, then stores the function.
# •	Implement dynamic dispatch using the registry.
# •	Support `*args` and `**kwargs` when invoking a registered operation.
# •	Raise ValueError for unsupported operation names.

def validate_num_args(*args):
    for value in args:
        if not isinstance(value, (int, float)) or isinstance(value,(bool)):
            raise TypeError(f"{value} is not a number type. Need numbers only.")

def add(*args):
    if len(args) != 2:
        raise ValueError(f"Expecting two arguments only.")
    
    validate_num_args(*args)
    
    return sum(args)

def subtract(*args):
    if len(args) != 2:
        raise ValueError(f"Expecting two arguments only.")
    
    validate_num_args(*args)
    
    return args[0] - args[1]

def multiply(*args):
    if len(args) != 2:
        raise ValueError(f"Expecting two arguments only.")
    
    validate_num_args(*args)
    
    return args[0] * args[1]

def divide(*args):
    if len(args) != 2:
        raise ValueError(f"Expecting two arguments only.")
    
    validate_num_args(*args)

    if args[1] == 0:
        raise ZeroDivisionError(f"Can't divide by Zero")
    
    return args[0] / args[1]


OPERATIONS = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply,
    "divide": divide
}

def register_operation(operation_name, operation):
    if operation_name is None:
        raise ValueError("No Operation Name was provided")
    
    if not isinstance(operation_name, str):
        raise TypeError("Operation Name has to be a Text or string value.")
    
    if not operation_name.strip():
        raise ValueError("Operation name cannot be empty.")

    if not callable(operation):
        raise TypeError("Not a callable operation or function")
    
    OPERATIONS[operation_name] = operation
    
    return f"{operation_name} for {operation} added to OPERATIONS registry."


def execute_operation(operation, *args, **kwargs):
    operation = OPERATIONS.get(operation_name)

    if operation is None:
        raise ValueError(f"No operation was provided in the argument.")

    if not callable(operation):
        raise ValueError(f"Unsupported Operation: {operation}")
    
    return operation(*args, **kwargs)
```

```python
# Second set for practice - 
def credit_account(balance, amount):
    if amount <= 0:
        raise ValueError("Credit Amount must be positive")
    return balance + amount

def debit_account(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient Fund")
    return balance - amount

def refund_account(balance, amount):
    if amount <= 0:
        raise ValueError("Credit Amount must be positive")
    return balance + amount

def calculate_transaction_fee(amount, rate):
    return amount * rate

def calculate_cashback(amount, rate):
    return amount * rate

TRANSACTION_REGISTRY = {
    "CREDIT": credit_account,
    "DEBIT": debit_account,
    "REFUND": refund_account,
    "FEE": calculate_transaction_fee,
}

def register_transaction(transaction_name, transaction):
    if not callable(transaction):
        raise TypeError("Operation must be callable")

    TRANSACTION_REGISTRY[transaction_name] = transaction

register_transaction(transaction_name="CASHBACK", transaction=calculate_cashback)

def execute_transaction(transaction_name, *args, **kwargs):
    transaction = TRANSACTION_REGISTRY.get(transaction_name)

    if transaction is None:
        raise ValueError(f"Unsupported operation: {transaction_name}")

    if not callable(transaction):
        raise TypeError(f"Registered Transaction Name is not callable: {transaction_name}")
    
    return transaction(*args, **kwargs)

print(execute_transaction("CREDIT", 1000, 500))
```

```python
# 1. CREDIT operations
assert execute_transaction("CREDIT", 1000, 500) == 1500
assert execute_transaction("CREDIT", 2000, 1000) == 3000

# 2. DEBIT operations
assert execute_transaction("DEBIT", 1000, 300) == 700
assert execute_transaction("DEBIT", 5000, 2000) == 3000

# 3. REFUND operations
assert execute_transaction("REFUND", 1000, 250) == 1250
assert execute_transaction("REFUND", 500, 100) == 600

# 4. FEE operations
assert execute_transaction("FEE", 1000, 0.02) == 20
assert execute_transaction("FEE", 5000, 0.01) == 50

# 5. Keyword argument execution
assert execute_transaction(
    "CREDIT", balance=1000, amount=500
) == 1500

assert execute_transaction(
    "DEBIT", balance=1000, amount=300
) == 700

assert execute_transaction(
    "REFUND", balance=1000, amount=250
) == 1250

assert execute_transaction(
    "FEE", amount=2000, rate=0.02
) == 40

# 6. Unsupported operation
try:
    execute_transaction("UNKNOWN", 100, 50)
except ValueError as error:
    assert "UNKNOWN" in str(error)
else:
    raise AssertionError("Expected ValueError")

# 7. Insufficient funds
try:
    execute_transaction("DEBIT", 100, 500)
except ValueError:
    pass
else:
    raise AssertionError("Expected ValueError")

# 8. Invalid credit amount
try:
    execute_transaction("CREDIT", 1000, -100)
except ValueError:
    pass
else:
    raise AssertionError("Expected ValueError")

# 9. Invalid refund amount
try:
    execute_transaction("REFUND", 1000, 0)
except ValueError:
    pass
else:
    raise AssertionError("Expected ValueError")

print("All Challenge 12 - Step 3 tests passed!")
```
