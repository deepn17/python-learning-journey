#### TUPLES — IMMUTABLE SEQUENCES
A tuple is like a list, but IMMUTABLE — you can't change it after creation. Written with parentheses (or just commas).

```python
# --- Creating tuples ---
point = (3, 4)             # a 2D coordinate
colors = ("red", "green", "blue")
single = (5,)              # ONE-item tuple needs a trailing comma!
not_a_tuple = (5)          # this is just the integer 5

print(point)
print(colors)
print(single)
print(not_a_tuple)
print(type(single))
print(type(not_a_tuple))
```

##### Why use tuples?
1. Immutability = safety. Data that shouldn't change (coordinates, days of the week, RGB values) is protected from accidental edits.
2. Tuples are slightly faster and use less memory than lists.
3. Tuples can be used as dictionary KEYS (lists cannot).

```python
DAYS = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
print(DAYS[0])                      # Mon  (indexing works the same)
```

```python
DAYS[0] = "Monday"
print(DAYS)
```

```python
# --- Tuple unpacking ---
# Unpacking assigns each element to its own variable in one line.
point = (3, 4)
x, y = point
print(f"x = {x}, y = {y}")
```

```python
# enumerate
for index, value in enumerate(["a", "b", "c"]):
    print(index, value)                          # unpacks (index, value) each time
```

```python
# Swapping two variables without a temp variable:
a, b = 1, 2
a, b = b, a
print(a, b)
```
