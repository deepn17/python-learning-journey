```python
# Challenge 3 — Transaction summary
#Turn a collection of transaction records into a compact financial summary that downstream reporting can consume.

# Business requirements
# •	Accept a collection of transactions.
# •	Classify transaction amounts as income or expense using the established transaction representation.
# •	Return total income, total expense, and net balance.
# •	Do not print the summary; return a structured result

# Technical requirements
# •	Implement `summarize_transactions(transactions)`.
# •	Iterate once where practical and accumulate income, expense, and balance.
# •	Define how transaction type and amount are represented; validate malformed records rather than guessing.
# •	Return a dictionary with stable keys such as `total_income`, `total_expense`, and `total_balance`.

def total_income(transaction):
    if not isinstance(transaction, (int, float)) or isinstance(transaction, (bool)):
        raise TypeError("Income has to be an Integer or float")
    if transaction < 0:
        raise ValueError("Income can't be Negative")
    return transaction

def total_expense(transaction):
    if not isinstance(transaction, (int, float)) or isinstance(transaction, (bool)):
        raise TypeError("Expense has to be an Integer or float")
    if transaction < 0:
        raise ValueError("Expense can't be Negative")
    return transaction

def summarize_transactions(transactions):
    income_total = 0
    expense_total = 0
    for transaction in transactions:
        if not isinstance(transaction, dict):
            raise TypeError("Transaction must be a dictionary")

        if 'type' not in transaction:
            raise ValueError("Missing required field: type")

        if 'amount' not in transaction:
            raise ValueError("Missing required field: amount")

        if not isinstance(transaction['type'], str):
            raise TypeError("Transaction type must be a string")

        transaction_type = transaction['type'].lower()

        if transaction_type == 'income':
            validated_income = total_income(transaction['amount'])
            income_total += validated_income
        
        elif transaction_type == 'expense':
            validated_expense = total_expense(transaction['amount'])
            expense_total += validated_expense
        
        else:
            raise ValueError("Unsupported transaction type")
    
    balance_total = income_total - expense_total

    return {
        "total_income": income_total,
        "total_expense": expense_total,
        "total_balance": balance_total
    }

        
```

```python
# TC - 1 Empty transactions
print(summarize_transactions([]))
```

```python
# TC 2 -  Mixed income and expense transactions
transactions = [
    {"type": "INCOME", "amount": 1000},
    {"type": "EXPENSE", "amount": 250},
    {"type": "INCOME", "amount": 500},
    {"type": "EXPENSE", "amount": 100},
    {"type": "INCOME", "amount": 200}
]

print(summarize_transactions(transactions))
```
