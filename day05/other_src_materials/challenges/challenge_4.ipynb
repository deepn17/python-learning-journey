```python
# Challenge 4 — Flexible transaction logger
# Create a flexible transaction logging function that accepts required positional data plus optional metadata, while preserving caller-owned metadata and applying safe defaults.

# Business requirements
# •	Capture transaction details supplied positionally and optional metadata supplied by keyword.
# •	Apply a default status when one is not provided.
# •	Support optional currency metadata and preserve unrelated metadata.
# •	Avoid unexpectedly changing a dictionary supplied by the caller.

# Technical requirements
# •	Implement `log_transaction(*args, **kwargs)`.
# •	Treat `kwargs` as the incoming metadata mapping.
# •	Copy metadata before applying defaults or modifications.
# •	Check for the presence of currency rather than assuming it exists.
# •	Return a predictable transaction record.


# Suggested implementation sequence
# .	Decide which positional arguments are expected and document their order.
# .	Capture optional metadata in kwargs.
# .	Create a copy before modifying metadata.
# .	Set status only when absent, not when explicitly supplied.
# .	Add or validate currency according to the contract.
# .	Return a new transaction record.

# Hints (not the full solution)
# •	`metadata = kwargs` refers to the same dictionary object; `metadata = kwargs.copy()` creates a shallow copy.
# •	Use `setdefault` when the requirement is “default only if key is absent”.
# •	A shallow copy is enough for top-level key updates; nested mutable values still share references.
# •	Avoid a variadic signature when a fixed signature would be clearer.
```

```python
def log_transaction(*args, **kwargs):
    transactions = args
    optional_metadata = kwargs

    metadata = optional_metadata.copy()
    metadata.setdefault("status", "PENDING")

    currency_exists = 'currency' in metadata.keys()


    return {
        "transaction_details": transactions,
        "metadata": metadata,
        "currency_exists": currency_exists
    }
```

```python
# TC 1
print(log_transaction('TXN001', 'payment', 10000, currency='INR', status='SUCCESS'))
```

```python
# TC 2
print(log_transaction('TXN002', 'transfer', 1000, currency='INR'))
```

```python
# TC 3
print(log_transaction('TXN001', 'payment', 10000, currency='INR', status=None))
```

```python
# TC 4 No metadata
result = log_transaction(101, 5000)
print(result)
```

```python
# TC 5 Explicit Status given
result = log_transaction(102, 3000, status="COMPLETED")
print(result)
```

```python
# TC 6 Currency supplied
result = log_transaction(103, 7000, currency="INR")
print(result)
```

```python
# TC 7 Currency and other metadata
result = log_transaction(
    104,
    9000,
    currency="INR",
    department="TREASURY",
    channel="ONLINE"
)
print(result)
```

```python
# TC 8 
original_metadata = {
    "currency": "INR",
    "department": "TREASURY"
}

result = log_transaction(101, 5000, **original_metadata)

print("Original metadata:", original_metadata)
print("Returned metadata:", result["metadata"])
```
