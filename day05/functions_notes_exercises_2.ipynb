```python
# Example 1 - Function without Parameters
def generate_full_name():
    first_name = "Kite"
    last_name = "Ranger"
    space = " "
    full_name = first_name + space + last_name
    print(full_name)

generate_full_name()
```

```python
# Example 1.1 - Function without Parameters
def add_two_numbers():
    num_one = 3
    num_two = 4
    total = num_one + num_two
    print(total)

add_two_numbers()
```

```python
# Example 1.2 - Function without Parameters and with Return value
def generate_full_name():
    first_name = "Kite"
    last_name = "Ranger"
    space = " "
    full_name = first_name + space + last_name
    return full_name

print(generate_full_name())
```

```python
# Example 1.3 - Function without Parameters and with Return value
def add_two_numbers():
    num_one = 3
    num_two = 4
    total = num_one + num_two
    return total

print(add_two_numbers())
```

```python
# Example 2 - Function with Parameters - One parameter
def greetings(name):
    message = name + ', Welcome to the Jungle '
    return message

print(greetings("You Can't See Me"))
```

```python
# Example 2.1 - Function with Parameters - One parameter
def add_ten(num):
    ten = 10
    return num + ten

print(add_ten(90))
```

```python
# Example 2.2 - Function with Parameters - One parameter - Squaring numbers
def square_number(x):
    return x * x

print(square_number(5))
```

```python
# Example 2.3 - Function with Parameters - One parameter - Area of circle
def area_of_circle(r):
    pi = 3.14
    area = pi * r ** 2
    return area

print(area_of_circle(10))
```

```python
# Example 2.4 - Function with Parameters - One parameter - Sum of numbers
def sum_of_numbers(n):
    total = 0
    for i in range(n + 1):
        total += i
    return total

print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

```python
# Example 3 - Function with Parameters - Two parameters
def generate_full_name(first_name, last_name):
    space = ' '
    full_name = first_name + space + last_name
    return full_name

print('Full Name: ', generate_full_name('Flying','Eagle'))

# Example 3.1 - Sum of 2 numbers
def sum_two_numbers(num_one, num_two):
    sum = num_one + num_two
    return sum

print('Sum of two numbers: ', sum_two_numbers(1, 9))

# Example 3.2 - Calculate Age
def calculate_age(current_year, birth_year):
    age = current_year - birth_year
    return age

print('Age: ', calculate_age(2021, 1819))

# Example 3.4 - Weight of Object in Newton's scale
def weight_of_object(mass, gravity):
    weight = str(mass * gravity)+ ' N'
    return weight
print('Weight of an object in Newtons: ', weight_of_object(100, 9.81))
```
