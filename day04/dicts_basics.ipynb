#### DICTIONARIES — KEY-VALUE PAIRS
A dictionary maps KEYS to VALUES, like a real dictionary maps words to definitions. Written with curly braces and colons. Keys must be unique and immutable (strings, numbers, tuples).

```python
# --- Creating and accessing ---
person = {
    "name": "Alice",
    "age": 30,
    "city": "Mumbai"
}
print(person["name"])
print(person["age"])
```
