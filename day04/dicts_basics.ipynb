#### DICTIONARIES — KEY-VALUE PAIRS
A dictionary maps KEYS to VALUES, like a real dictionary maps words to definitions. Written with curly braces and colons. Keys must be unique and immutable (strings, numbers, tuples).

```python
# --- Creating and accessing ---
person = {
    "name": "Alice",
    "age": 30,
    "city": "Mumbai"
}
print(person["name"])
print(person["age"])
```

```python
# --- Adding, updating, removing ---
person = {"name": "Alice","age": 30}

person["job"] = "Engineer"  #   Add a new key
person["age"] = 31          #   update an existing key

print(f"Dictionary after adding a new key and updating an existing key: {person}")

del person["job"]

print(f"Dictionary after deleting a key: {person}")
```

```python
# --- get() — the safe accessor / get() vs []---
person = {"name": "Alice"}

print(person["age"])         # Using [] on a missing key raises a KeyError: print(person["age"])   # KeyError: 'age'
```

```python
print(person.get("age"))           # None
print(person.get("age", 0))        # 0  (custom default)
# get() returns None (or a default) instead of crashing
```

```python
# --- Looping over dicts ---
person = {
    "name": "Alice",
    "age": 30,
    "city": "Mumbai"
}

for key in person:          # loops over KEYS by default
    print(key)

for key, value in person.items():  # keys AND values together
    print(f"{key} → {value}")

print(person.keys())
print(person.values())
```

```python
# --- Membership ---
person = {"name": "Alice", "age": 30}
print("name" in person)     #True
print("Alice" in person)    #False, Alice is a value and not a key.
```

```python
# Count frequencies with a dict ---
text = "the quick brown fox jumps over the lazy dog"
freq = {}

for word in text.split():
    freq[word] = freq.get(word, 0) + 1
print(freq)
```
