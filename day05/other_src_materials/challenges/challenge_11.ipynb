```python
# Challenge 11 — Function aliases, rebinding, equality, and identity
# Understand that functions are objects and that assigning a function to another variable creates another reference to the same object—not a new function.

# Business requirements
# •	Allow a function to be referenced by more than one variable.
# •	Demonstrate what happens when one name is rebound to a different function.
# •	Distinguish equality (`==`) from identity (`is`).
# •	Reject non-callable values when an API requires a function.

# Technical requirements
# •	Create aliases by assignment, e.g. `calculate_total_alias = calculate_total`.
# •	Verify that both names initially refer to the same object.
# •	Rebind `calculate_total` and verify the alias still refers to the original function.
# •	Use `is` for object identity; do not use `==` as a substitute.
# •	Raise TypeError when a supplied operation is not callable.

def validate_num_args(*args):
    for values in args:
        if not isinstance(values, (int, float)) or isinstance(values, bool):
            raise TypeError("Expecting numerical values")

def calculate_total(*args):
    validate_num_args(*args)
    return sum(args)

def check_identity_function(func1, func2):
    if not callable(func1) or not callable(func2):
        raise TypeError("Not a callable function")
    if func1 is func2:
        return True
    else:
        return False

calculate_total_alias = calculate_total


def execute_operation(operation, *args):
    if not callable(operation):
        raise TypeError("Not a callable function")
    return operation(*args)

print(execute_operation(calculate_total, 10, 20))
print(execute_operation(calculate_total_alias, 10, 20))

# REBINDING
def calculate_total(*args):
    return max(args)


print(calculate_total_alias(10, 20, 30))
print(calculate_total(10, 20, 30))

print(calculate_total_alias is calculate_total)
print(calculate_total_alias == calculate_total)


new_alias = calculate_total

print(new_alias(10, 20, 30))
print(new_alias is calculate_total)
print(new_alias is calculate_total_alias)
```
