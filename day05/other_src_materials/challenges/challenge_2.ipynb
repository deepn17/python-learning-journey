```python
# Challenge 2 — Higher-order calculator
# Create reusable arithmetic operations and a calculator that receives the selected operation as a function, rather than hardcoding every branch into one large function.

# Business requirements
# •	Support addition, subtraction, multiplication, and division.
# •	Allow a caller to select the operation.
# •	Provide a clear failure for division by zero.
# •	Reject an unsupported operation rather than silently returning a misleading result.

# Technical requirements
# •	Implement independent functions such as `add`, `subtract`, `multiply`, and `divide`.
# •	Implement `calculate(a, b, operation)` where operation is callable or is resolved through a mapping.
# •	Validate the operation before invocation.
# •	The recorded exercise used the message `Can’t Divide by Zero` for the zero-division case; preserve the agreed contract in your tests.

# Notes
# •	Functions are first-class objects
# •	Higher-order functions
# •	Callable validation
# •	Dictionary-based dispatch
# •	Single-responsibility design
```

```python
# step 1 Define each arithmetic function independently
def add(a, b):
    """
    Calculate the sum of two numeric operands

    Business Purpose:
        Support addition as a reusable arithmetic operation.

    Parameters:
        a (int | float): First Operand
        b (int | float): Second Operand

    Returns:
        int | float: sum of the two operands
    """
    return a + b

def subtract(a, b):
    """
    Calculate the difference of two numeric operands

    Business Purpose:
        Support subtraction as a reusable arithmetic operation.

    Parameters:
        a (int | float): First Operand
        b (int | float): Second Operand
    
    Returns:
        int | float: Difference of two operands
    """
    return a - b

def multiply(a, b):
    """
    Calculate the product of two numeric operands

    Business Purpose:
        Support multiplication as a reusable arithmetic operation.

    Parameters:
        a (int | float): First Operand
        b (int): Second Operand
    
    Returns:
        int: Product of two operands
    """
    return a * b


def divide(a, b):
    """
    Divide the first numeric operand by the second.

    Business Purpose:
        Support division while preventing invalid zero-divisor operations.

    Parameters:
        a (int): Dividend (the number being divided).
        b (int): Divisor (the number by which a is divided).
    
    Returns:
        float: Quotient resulting from the division.
    
    Raises:
        ZeroDivisionError: If the divisor is zero.
    """
    if b == 0:
        raise ZeroDivisionError("Can’t Divide by Zero")
    return a / b

# STEP 2: Map operation names to their respective function references.

OPERATIONS = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply,
    "divide": divide,
}

# STEP 3: Implement the higher-order calculator.
def calculate(a, b, operation):
    """
    Purpose:
        Execute a supplied arithmetic function using two operands.

    Business Purpose:
        Allow callers to select an arithmetic operation without
        hardcoding individual operations inside the calculator.

    Parameters:
        a (int | float): First operand.
        b (int | float): Second operand.
        operation (callable): Arithmetic function to execute.

    Returns:
        Any: Result returned by the supplied operation.

    Raises:
        TypeError: If the supplied operation is not callable.
        Exception: Any exception raised by the supplied operation
                   is propagated to the caller.

    Design Decisions:
        Uses a higher-order function because it accepts another
        function as an argument.
        Uses callable() to validate the supplied operation.
        Does not catch exceptions raised by the operation.
    """
    if not callable(operation):
        raise TypeError(f"Operation '{operation}' must be callable")

    return operation(a, b)
    
```
